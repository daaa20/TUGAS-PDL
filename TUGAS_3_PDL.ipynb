```python
%pip install pandas Sastrawi

import pandas as pd
import re
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
```

```python
teks = "Mahasiswa perlu memanfaatkan teknologi untuk mendukung proses pembelajaran. Penggunaan internet memudahkan mahasiswa dalam mencari informasi dan memahami materi perkuliahan. Selain itu, kemampuan mengolah data menjadi keterampilan penting di era digital. Oleh karena itu, mahasiswa harus terus belajar agar mampu menghadapi perkembangan teknologi."

data = pd.DataFrame({"teks": [teks]})
print(data["teks"][0])
```

```python
data["huruf_kecil"] = data["teks"].str.lower()
data["bersih"] = data["huruf_kecil"].str.replace(r"[^a-z0-9\\s]", " ", regex=True)
data["bersih"] = data["bersih"].str.replace(r"\\s+", " ", regex=True).str.strip()

print("Teks asli:")
print(data["teks"][0])
print("\\nHasil case folding dan pembersihan tanda baca:")
print(data["bersih"][0])
```

```python
contoh = data.loc[0, "bersih"]
token = contoh.split()

print("Teks:", contoh)
print("Token:", token)
print("Jumlah token:", len(token))
```

```python
stopwords = set(StopWordRemoverFactory().get_stop_words())
stopwords = stopwords - {"tidak", "bukan", "jangan", "belum"}

token_penting = []
for kata in token:
    if kata not in stopwords:
        token_penting.append(kata)

print("Sebelum penghapusan stopwords:")
print(token)
print("\\nSesudah penghapusan stopwords:")
print(token_penting)
print("Jumlah token sebelum:", len(token))
print("Jumlah token sesudah:", len(token_penting))
```

```python
stemmer = StemmerFactory().create_stemmer()
hasil_stem = [stemmer.stem(kata) for kata in token_penting]

print("Sebelum stemming:", token_penting)
print("Sesudah stemming:", hasil_stem)
```

```python
data["tokenisasi"] = [token]
data["tanpa_stopwords"] = [token_penting]
data["stemming"] = [hasil_stem]

pd.set_option("display.max_colwidth", None)
display(data[["teks", "bersih", "tokenisasi", "tanpa_stopwords", "stemming"]])
```
